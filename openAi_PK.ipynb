```python
# real time promt engerninng
from IPython.display import HTML
def get_compleation(prompt, model="o4-mini"):
  messages=[{"role":"user","content":prompt}]
  responses=client.chat.completions.create(model=model,messages=messages,temperature=1)
  return responses.choices[0].message.content
```

```python
from IPython.display import Markdown, display
chat=get_compleation(prompt="write a story about sea",model="o4-mini")
display(Markdown(chat))

```
